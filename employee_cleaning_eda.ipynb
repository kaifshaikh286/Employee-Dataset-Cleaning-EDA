```python
df=df.drop_duplicates()
df.shape
df.duplicated().sum()
```

```python
df['Age'].median()
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
df['Age'].isnull().sum()
```

```python
df['Department']=df['Department'].fillna('Unknown')
df['Department'].isnull().sum()
```

```python
df['City']=df['City'].fillna('Unknown')
df['City'].isnull().sum()
```

```python
df['Salary'].median()
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].median())
df['Salary'].isnull().sum()
```

```python
df['YearsExperience'].median()
```

```python
df['YearsExperience']=df['YearsExperience'].fillna(df['YearsExperience'].median())
df['YearsExperience'].isnull().sum()
```

```python
df['EmploymentStatus']=df['EmploymentStatus'].fillna('Unknown')
df['EmploymentStatus'].isnull().sum()
```

```python
df['Education']=df['Education'].fillna('Unknown')
df['Education'].isnull().sum()
```

```python
df['PerformanceScore'].median()
```

```python
df['PerformanceScore']=df['PerformanceScore'].fillna(df['PerformanceScore'].median())
df['PerformanceScore'].isnull().sum()
                        
```

```python
df['Email']=df['Email'].fillna('Not Provided')
df['Email'].isnull().sum()
```

```python
df['JoiningDate']=pd.to_datetime(df['JoiningDate'],dayfirst=True,format='mixed',errors='coerce')
df['JoiningDate'].isnull().sum()
df=df.dropna(subset=['JoiningDate'])
print(df['JoiningDate'].isnull().sum())
```

```python
df
```

```python
df['FullName']=df['FullName'].str.title()
df['FullName'].value_counts()
```

```python
df['Department']=df['Department'].str.title()
df['Department'].value_counts()
```

```python
df['City']=df['City'].str.strip().str.title()
df['City'].value_counts()
```

```python
df['EmploymentStatus']=df['EmploymentStatus'].str.strip().str.title()
df['EmploymentStatus'].value_counts()
```

```python
df['Education']=df['Education'].str.strip().str.title()
df['Education'].value_counts()
```

```python
df['Department']=df['Department'].replace('It','IT')
df['Department'].value_counts()
```

```python
df['Department']=df['Department'].replace('Hr','HR')
df['Department'].value_counts()
```

```python
df['Education']=df['Education'].replace('Phd','PhD')
df['Education'].value_counts()
```

```python
(df['Salary'] < 0).sum()   # check how many negative values exist
```

```python
df['Salary'] = df['Salary'].abs()
(df['Salary'] < 0).sum()   # should now be 0
df['Salary'].min()          # should now show a positive number
```

```python
import matplotlib.pyplot as plt
dept_salary=df.groupby('Department')['Salary'].mean().sort_values(ascending=False).round(4)
plt.figure(figsize=(8,5))
dept_salary.plot(kind='bar', color='seagreen')
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('avg_salary_by_department.png', bbox_inches='tight')
plt.show()
```

```python
status_counts = df['EmploymentStatus'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(status_counts, labels=status_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Employee Status Distribution')
plt.savefig('employment_status_distribution.png', bbox_inches='tight')
plt.show()
```
